# Part A — Decoder-Only Model

## Decoder-only example: GPT-style Text Generation

A decoder-only model predicts the next token based on previous tokens.

We will use **DistilGPT-2**, a small GPT-style model.


````python
generator = pipeline("text-generation", model="distilgpt2")

prompt = "Natural language processing is"

output = generator(prompt, max_new_tokens=30, do_sample=False)

print(output[0]["generated_text"])
````

# Part B — Encoder–Decoder Model

## Encoder–decoder example: Instruction / Text-to-Text Generation

Encoder–decoder models contain two main parts:

1. The **encoder** reads the input.
2. The **decoder** produces the output.

We will use **FLAN-T5-small**.


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

prompt = "Translate to French: I love natural language processing."

# Convert text into tokens
output = model.generate(**tokenizer(prompt, return_tensors="pt"))

# Convert tokens back to text
print(tokenizer.decode(output[0], skip_special_tokens=True))
````

# Part C — Encoder-Only Model

## Encoder-only example: Sentiment Classification

Encoder-only models read the full input and create a representation of it.

They are very useful when the task is to **understand or classify text**.

Here we use a DistilBERT sentiment model.


````python
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

text = "This NLP lab is easy and interesting."

result = classifier(text)
print(result)
````

## Task 2 — Encoder-only Model

Use the sentiment classifier on these three sentences:

```text
The movie was excellent.
The lecture was difficult to understand.
I enjoyed learning about transformers.
```

Print the predicted label and score for each sentence.

````python
# Task 2 - Encoder-only Model
from transformers import pipeline

classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

sentences = [
    "The movie was excellent.",
    "The lecture was difficult to understand.",
    "I enjoyed learning about transformers.",
]

for sentence in sentences:
    result = classifier(sentence)[0]
    print(f"{sentence!r} -> label: {result['label']}, score: {result['score']:.4f}")
````

## Task 3 — Encoder–Decoder Model

Use the FLAN-T5 model to perform the following:

- Summarize a short paragraph.


````python
# Task 3 - Encoder-Decoder Model: summarize a short paragraph
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

summ_tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
summ_model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

paragraph = (
    "Natural language processing (NLP) is a field of artificial intelligence that focuses "
    "on enabling computers to understand, interpret, and generate human language. It combines "
    "computational linguistics with machine learning and deep learning models. Common NLP "
    "applications include machine translation, sentiment analysis, chatbots, and text "
    "summarization. Modern NLP systems rely heavily on transformer-based architectures such as "
    "BERT, T5, and GPT."
)

prompt = "Summarize: " + paragraph
inputs = summ_tokenizer(prompt, return_tensors="pt")
summary_ids = summ_model.generate(**inputs, max_new_tokens=60)

print("Original paragraph:\n", paragraph)
print("\nSummary:\n", summ_tokenizer.decode(summary_ids[0], skip_special_tokens=True))
````


## Task 5 — Short Reflection

Answer in 3–5 sentences:

**Why is a decoder-only model suitable for a chatbot such as ChatGPT?**


**Answer:** A decoder-only model is suitable for a chatbot like ChatGPT because it is trained to predict the next token given everything said so far, which is exactly the structure of an open-ended conversation: the growing dialogue history (system prompt, user turns, and the model's own previous replies) is just one long sequence the model keeps extending. This autoregressive, single-stream design lets it generate fluent, variable-length, open-ended responses token by token without needing a separate fixed-size input to encode first, unlike encoder-only models (which only classify/understand text and don't generate) or encoder-decoder models (which are built around transforming one fixed input sequence into one output sequence, e.g. translation, rather than an ever-growing multi-turn conversation).
